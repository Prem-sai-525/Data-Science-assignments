ESTIMATION AND CONFIDENCE INTERVALS

a. Build 99% Confidence Interval Using Sample Standard Deviation        
Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.


```python
import numpy as np
from scipy import stats
```

```python
data=[1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
#calculate sample size and sample mean
n=len(data)
mean=np.mean(data)
print(f"sample size= {n}, mean= {mean:.4f}")
```

```python
#calculate sample standard deviation
sample_std=np.std(data,ddof=1)
print(f"sample standard deviation:{sample_std:.4f}")
```

```python
#building 99% confidence interval
t_ci=stats.t.interval(0.99,df=n-1,loc=mean,scale=sample_std/np.sqrt(n))
print(f"(t-distribution) 99% CI: [{t_ci[0]:.4f},{t_ci[1]:.4f}]")
```

b. Build 99% Confidence Interval Using Known Population Standard Deviation      
If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.


```python
pop_mean=0.2
```

```python
#building 99% confidence interval using known population standard deviation
t_ci=stats.t.interval(0.99,df=n-1,loc=mean,scale=pop_mean/np.sqrt(n))
print(f"((t-distribution) 99% CI: [{t_ci[0]:.4f},{t_ci[1]:.4f}])")
```

HYPOTHESIS TESTING

1. State the Hypotheses statement:

```python
print("H0: There is no increase in weekly operating costs")
print("H1: There is increase in weekly operating costs")
```

2. Calculate the Test Statistic:

Use the following formula to calculate the test statistic(z):

where:

●	ˉxˉ = sample mean weekly cost (Rs. 3,050)

●	μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)

●	σ = 5*25 units

●	n = sample size (25 restaurants)


```python
import math
import scipy.stats
```

```python
sample_mean=3050
mean_units=600
std_units=25
n=25
```

```python
#calculate theoritical mean
mu=1000+(5*mean_units)
```

```python
#calculate standard deviation
sigma=5*std_units
```

```python
#calculate standard error
se=sigma/math.sqrt(n)
```

```python
#calculate z_test
z=(sample_mean-mu)/(se)
```

```python
print("Theoritical mean=",mu)
print("Standard deviation=",sigma)
print("Standard error=",se)
print("Test statistic(z)=",z)
```

3. Determine the Probability and compare:   
Using the alpha level of 5% (α = 0.05),


```python
alpha=0.05
```

```python
from scipy.stats import norm
```

```python
#calculate p value
p_value=1-norm.cdf(z)
```

```python
#calculate critical value
critical_value=norm.ppf(1-alpha)
```

```python
print("P-value=",p_value)
print("Critical value=",critical_value)
```

4. Make a Decision:     
Compare the test statistic with the critical value to decide whether to reject the null hypothesis.


```python
if z>critical_value:
    print("Reject null hypothesis(H0)")
else:
    print("failed to reject null hypothesis(H0)")
```
