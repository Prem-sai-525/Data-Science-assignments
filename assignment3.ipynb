Descriptive Analytics for Numerical Columns

●	Objective: To compute and analyze basic statistical measures for numerical columns in the dataset.

●	Steps:

	Load the dataset into a data analysis tool or programming environment (e.g., Python with pandas library).

	Identify numerical columns in the dataset.

	Calculate the mean, median, mode, and standard deviation for these columns.
    
	Provide a brief interpretation of these statistics.


```python
#Load the dataset into a data analysis tool or programming environment 
df=pd.read_csv('sales_data_with_discounts.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
#Identify numerical columns in the dataset.
num_cols=df.select_dtypes(include='number').columns
```

```python
print(num_cols)
```

```python
#Calculate the mean, median, mode, and standard deviation for these columns.
for col in num_cols:
    print(f'\n{col}')
    print(f'Mean:',df[col].mean())
    print(f'Median:',df[col].median())
    print(f'Mode:',df[col].mode()[0])
    print(f'Standard Deviation:',df[col].std())
```

```python
#Provide a brief interpretation of these statistics.
print(
""" 
Volume: The average volume is about 5 units, while the median is 4 units. \nThis means that most sales are for a small number of units. \nThe values have some variation, as shown by the standard deviation of 4.23.
Avg Price: The average price is 10,453.43, but the median is only 1,450. \nThis shows that a few products have very high prices, which increases the average.
Total Sales Value: The average total sales value is 33,812.84, while the median is 5,700. \nThis means that some transactions have much higher sales values than others.\n
Discount Rate: The average discount rate is around 15.16%, and the median is 16.58%.\n Since these values are fairly close, the discount rates are relatively consistent.
\nDiscount Amount: The average discount amount is 3,346.50, while the median is 988.93. \nThis suggests that some customers or products have received much larger discounts than others.
Net Sales Value: The average net sales value is 30,466.34, compared with a median of 4,677.79. \nThis indicates that a few sales have very high net values, causing the average to increase.

""")
```

```python
#Plot histograms for each numerical column.
df[num_cols].hist(figsize=(10,12),bins=20, edgecolor='black')

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
#Analyze the distribution (e.g., skewness, presence of outliers) and provide inferences.
print(
"""
The histograms show that Volume is mostly concentrated at lower values, with a few higher values.\n Avg Price and Total Sales Value are strongly right-skewed because a few products or transactions have very high values. \n Discount Rate (%) is more evenly distributed, with most values falling between 5% and 20%. \nDiscount Amount and Net Sales Value are also right-skewed, as most values are relatively small but a few are much larger. \nOverall, the numerical variables show considerable variation and some possible outliers, especially in the sales and price-related columns.

""")
```

```python
#Create boxplots for numerical variables to identify outliers and the interquartile range.
plt.figure(figsize=(10,5))
df[num_cols].boxplot()
plt.show()
```

```python
def outlier_capping(df,col):
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1
    lower_extreme=Q1-1.5*IQR
    upper_extreme=Q3-1.5*IQR
    df[col]=df[col].apply(lambda x: lower_extreme if x< lower_extreme else upper_extreme if x>upper_extreme else x)

for col in num_cols:
    outlier_capping(df,col)
```

```python
#Discuss any findings, such as extreme values or unusual distributions.
print(
"""
The box plots show that Volume has a few higher values that can be considered outliers. \nAvg Price has several extreme high values, showing that some products are much more expensive than others. \nTotal Sales Value and Net Sales Value also contain some high-value outliers, which means a few transactions have much larger sales than the majority. \nDiscount Amount has some extreme values as well, while Discount Rate (%) is comparatively more evenly distributed with fewer extreme values.

"""
)
```

```python
#Identify categorical columns in the dataset.
cat_cols = df.select_dtypes(include=['object']).columns
cat_cols
```

```python
#Create bar charts to visualize the frequency or count of each category.
for col in cat_cols:
    plt.figure(figsize=(10,5))

    counts=df[col].value_counts()

    plt.bar(counts.index,counts.values)
    plt.title("Bar Chart of " + column)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
print(
""" The bar charts show how frequently each category appears in the dataset. \nBusiness Unit is evenly distributed, with the three business units having a similar number of records. \nThe Day column is also fairly balanced, although Thursday has slightly more records than the other days.\n The SKU and Model categories have a relatively balanced distribution. The Brand column has some differences in frequency, meaning certain brands appear more often than others. \nThe City column contains only one category, so there is no variation between cities in this dataset."""
)
```
