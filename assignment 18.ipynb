```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

```python
lab_enc=LabelEncoder()
std_sca=StandardScaler()
```

```python
x=std_sca.fit_transform(x)
y=lab_enc.fit_transform(y)
```

2. Model Implementation   
●	Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.    
●	Divide the dataset into training and test sets.   
●	Train your model on the training set and then use it to make predictions on the test set.


```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
model=Sequential()
```

```python
model.add(Dense(units=12,activation='relu',input_shape=(60,))) #first hidden layer
model.add(Dense(units=1,activation='sigmoid')) #output layer
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=50,batch_size=16)
```

```python
pred=model.predict(x_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,pred.round(),target_names=['Rock','Mine']))
```

3. Hyperparameter Tuning  
●	Modify various hyperparameters, such as the number of hidden layers, neurons per hidden layer, activation functions, and learning rate, to observe their impact on model performance.   
●	Adopt a structured approach like grid search or random search for hyperparameter tuning, documenting your methodology thoroughly.


```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid={
    'hidden_layer_sizes':[(12,),(24,)],
    'activation':['relu','tanh'],
    'learning_rate_init':[0.001,0.01]
}
```

```python
grid_search=GridSearchCV(estimator=model,param_grid=param_grid,cv=3,scoring='accuracy')
```

```python
from sklearn.neural_network import MLPClassifier
```

```python
mlp=MLPClassifier(max_iter=500,random_state=42)
```

```python
grid_search=GridSearchCV(estimator=mlp,param_grid=param_grid,cv=3,scoring='accuracy')
```

```python
grid_search.fit(x_train,y_train)
```

```python
grid_search.best_params_,grid_search.best_score_
```

```python
best_model=grid_search.best_estimator_
```

```python
tuned_pred=best_model.predict(x_test)
```

4. Evaluation   
●	Employ suitable metrics such as accuracy, precision, recall, and F1-score to evaluate your model's performance.   
●	Discuss the performance differences between the model with default hyperparameters and the tuned model, emphasizing the effects of hyperparameter tuning.


```python
print(classification_report(y_test,tuned_pred,target_names=['Rock','Mine']))
```
