1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
lis=[1,2,3,4,5]
print(lis)
```

```python
#extend
lis1=[9,7,6]
lis.extend(lis1)
print(lis)
```

```python
#insert
lis.insert(4,66)
print(lis)
```

```python
#remove
lis.remove(6)
print(lis)
```

```python
#pop
lis.pop(1)
print(lis)
```

```python
#clear
temp=lis.copy() #copied the list to a variable named temp
print(temp)
temp.clear()
print(temp)
```

```python
#count
c=lis.count(66)
print(c)
```

```python
#sort
lis.sort()
print(lis)
```

```python
#reverse
lis.reverse()
print(lis)
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
num=10
if num>0:
    if num%2==0:
        print(f"{num} is positive and also even")
    else:
        print(f"{num} is positive but not even")
elif num==0:
    print(f"{num} is zero")
else:
    print(f"{num} is negative")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
# Creating dataframe with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'
data = {
    'Name': ['Tony Stark', 'Bruce Wayne', 'Peter Parker', 'Clark Kent', 'Diana Prince', 
             'Barry Allen', 'Natasha Romanoff', 'Steve Rogers', 'Matt Murdock', 'Peter Parker'],
    'Age': [45, 38, 21, 33, 800, 28, 34, 105, 30, 21],
    'City': ['New York', 'Gotham', 'Queens', 'Metropolis', 'Washington D.C.', 
             'Central City', 'Moscow', 'Brooklyn', "Hell's Kitchen", 'Queens'],
    'Salary': [5000000, 5000000, 35000, np.nan, 95000, 72000, 120000, np.nan, 55000, 35000]
}
```

```python
#Converting to DataFrame
df = pd.DataFrame(data)
```

```python
df.info()
```

```python
df.describe()
```

```python
#Select 'Name' and 'City' columns
df[['Name','City']]
```

```python
#Drop 'City' column
df.drop(columns=['City'],inplace=True)
df
```

```python
#Fill any missing values in 'Salary' column with the mean
df.isnull().sum()
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
```

```python
df
```

```python
#Remove any duplicate rows
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```
